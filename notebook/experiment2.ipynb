```python
from langchain_core.tools import tool
from typing import List
from dotenv import load_dotenv
from debugging.exception import customException
from groq import Groq
import requests
import os
import re
import sys

# Load environment variables from .env file
load_dotenv()

# Load Google search credentials
api_key = os.getenv('GOOGLE_SEARCH_API_KEY')
cse_id = os.getenv('CSE_ID')

# Initialize the Groq client with API key
client = Groq(api_key=os.environ.get("GROQ_API_KEY"))

# Helper function
@tool
def fetch_google_images(query: str) -> List[tuple[str,str]]:
    """
    This is the 'fetch_google_images' tool, designed to retrieve and analyze images 
    based on a user-provided query topic. It is tightly scoped for use in specific, 
    visual exploration tasks.

    ✅ Usage Conditions:
    - Use only when the user explicitly asks to see images or visual examples related to a topic.
    - Use when the LLM determines that showing relevant images would significantly enhance 
      understanding (e.g., visual explanation, comparisons, or illustration).
    - Always ensure the user query has a clear and meaningful topic.

    ❌ Do NOT use this tool:
    - When the user query is ambiguous or off-topic.
    - To guess or infer unrelated images.
    - As a default fallback — it must be purpose-driven and user-relevant.

    🔍 What it does:
    - Searches Google for top relevant images matching the query.
    - Analyzes each image using the Groq API to extract detailed visible content (up to 512 tokens).
    - Returns a list of (image URL, image summary) pairs.

    Parameters:
    ----------
    query : str
        The search term or topic to fetch and analyze images for.

    Returns:
    -------
    List[tuple[str, str]]
        A list of tuples, each containing:
        - The image URL (str).
        - A detailed Groq-generated summary of the image's visual content (str).

    Raises:
    ------
    customException
        If there is a failure in fetching or analyzing images.

    Notes:
    -----
    - This tool combines Google image search with rich visual interpretation using Groq.
    - Intended for use in agent-based systems like LangGraph to deliver enhanced multimodal responses.
    - Use only in scenarios where visual context is directly beneficial or explicitly requested.
    """
    try:
        # Base URL for Google Custom Search API
        url = 'https://www.googleapis.com/customsearch/v1'
        
        # Exclude social media platforms in search results
        excluded_sites = '-site:instagram.com -site:facebook.com -site:twitter.com'
        modified_query = f"{query} {excluded_sites}"

        # Build request parameters for Google Custom Search API
        params = {
            'q': modified_query,
            'cx': cse_id,
            'key': api_key,
            'num': 1,
            'searchType': 'image',
            'safe': 'medium'  # Safe search level
        }

        # Send the request to Google API to retrieve images
        response = requests.get(url, params=params,verify=True)
        data = response.json()

        images = []
        if 'items' in data:
            for item in data['items']:
                images.append(item['link'])

        # Initialize list to store (image_url, summary) pairs
        list_desc = []

        # For each image, call Groq API to analyze and summarize
        for i in images:
            completion = client.chat.completions.create(
                model="meta-llama/llama-4-scout-17b-16e-instruct",
                messages=[
                    {
                        "role": "user",
                        "content": [
                            {
                                "type": "text",
                                "text": "Extract all key entities and important details from this image. Keep it brief but complete."
                            },
                            {
                                "type": "image_url",
                                "image_url": {
                                    "url": i
                                }
                            }
                        ]
                    }
                ],
                temperature=1,
                max_completion_tokens=None,
                top_p=1,
                stream=False,
                stop=None,
            )

            # Append (image_url, summary) to the list
            list_desc.append((i, completion.choices[0].message.content))
        if list_desc:
            return list_desc
        return ['''Sorry I Couldn't find any image related to this query please don't hallucinate just say 'I'm sorry i couldn't find [Topic]''']

    except Exception as e:
        # Raise custom exception for any error
        raise customException(e, sys)
```

```python
from langgraph.graph import StateGraph, END, MessagesState
from langgraph.prebuilt import ToolNode

from utils.analyze_image import analyze_image
from utils.google_image_fetching_tool import fetch_google_images

from langchain_core.messages import SystemMessage, HumanMessage
from langchain.memory import ConversationBufferWindowMemory
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv
import re
# Load environment variables
load_dotenv()

# Initialize conversation memory with a window of 35 messages
memory = ConversationBufferWindowMemory(k=35, return_messages=True)

# # Load system prompt template
# with open(r"E:\BegetBiz_BOT\Agents\system_prompt.txt", 'r') as f:
#     template = f.read()

# Register tools
tools = [fetch_google_images]

# Bind tools to async-compatible LLM
llm_with_tools=ChatOpenAI(model='gpt-4o-mini',).bind_tools(tools)

memory.chat_memory.add_message(SystemMessage(content=template))
def AGENT() -> StateGraph:
    # Utility to extract image URL
    def extract_image_url(text):
        pattern = r'https?://[^\s]+?\.(jpg|jpeg|png|gif|webp)'
        match = re.search(pattern, text, re.IGNORECASE)
        return match[0] if match else None

    # Async model call
    async def call_model(state: MessagesState) -> MessagesState:
        messages = state['messages']
        user_query = messages[0]
        modified_messages = messages.copy()

        image_url = extract_image_url(user_query.content)
        if image_url:
            # Await the image analysis tool
            image_description = await analyze_image.ainvoke({'url': image_url})
            modified_messages[0] = HumanMessage(
                content=f'''
Please reference this image description and based on it I will ask you a query.
<image_description>
{image_description}
</image_description>
Image Link: {image_url}
NOTE : If you get 'URL is invalid or Failed to analyze image' in place of description feel free to say me that 'URL is invalid'
                '''
            )

        # Load memory and merge
        chat_history = memory.load_memory_variables({})['history']
        recent_state_with_history = chat_history + modified_messages

        # LLM call
        response = await llm_with_tools.ainvoke(recent_state_with_history)

        # ✅ Final
        return {"messages": [response]}


    # Router node
    def router_node(state: MessagesState) -> str:
        last_message = state['messages'][-1]
        if last_message.tool_calls:
            return 'continue'
        else:
            memory.chat_memory.add_messages(state['messages'])
            return 'respond'

    # Create async-compatible workflow
    workflow = StateGraph(MessagesState)
    workflow.add_node("llm", call_model)
    workflow.add_node("tools", ToolNode(tools))
    workflow.set_entry_point("llm")
    workflow.add_conditional_edges(
        "llm",
        router_node,
        {
            "respond": END,
            "continue": "tools"
        }
    )
    workflow.add_edge("tools", "llm")

    return workflow.compile()

```

```python
from langgraph.graph import StateGraph, END, MessagesState
from langgraph.prebuilt import ToolNode

from utils.analyze_image import analyze_image
from utils.google_image_fetching_tool import fetch_google_images

from langchain_core.messages import SystemMessage, HumanMessage
from langchain.memory import ConversationBufferWindowMemory
from langchain_groq import ChatGroq

from dotenv import load_dotenv
import re

# Load environment variables
load_dotenv()

# Initialize conversation memory with a window of 35 messages
memory = ConversationBufferWindowMemory(k=35, return_messages=True)

# Load system prompt template
with open(r"E:\BegetBiz_BOT\Agents\system_prompt.txt", 'r') as f:
    template = f.read()

# Register tools
tools = [fetch_google_images]

# Bind tools to LLM
llm_with_tools = ChatGroq(model='llama3-70b-8192', temperature=0.1).bind_tools(tools)



# Utility function to extract an image URL from text
def extract_image_url(text):
    pattern = r'https?://[^\s]+?\.(jpg|jpeg|png|gif|webp)'
    match = re.search(pattern, text, re.IGNORECASE)
    return match[0] if match else None

# Core function to call the model
def call_model(state: MessagesState) -> MessagesState:
    messages = state['messages']
    user_query = messages[0]

    # Clone the messages list for safe mutation
    modified_messages = messages.copy()

    # Check for image URL and analyze if present
    image_url = extract_image_url(user_query.content)
    if image_url:
        image_description = analyze_image.invoke({'url': image_url})
        modified_messages[0] = HumanMessage(
            content=f'''
Please reference this image description and based on it I will ask you a query.
<image_description>
{image_description}
</image_description>
Image Link: {image_url}
NOTE : If you get 'URL is invalid or Failed to analyze image' in place of description feel free to say me that 'URL is invalid'
            '''
        )

    print("modified_messages: ",modified_messages)

    # Load memory history and merge with modified current messages
    chat_history = memory.load_memory_variables({})['history']
    recent_state_with_history = [SystemMessage(content=template)] + chat_history + modified_messages
    print('recent_state_with_history :',recent_state_with_history)
        
    # Invoke the LLM with tools
    response = llm_with_tools.invoke(recent_state_with_history)

    # Save to memory if no tools are used
    if not response.tool_calls:
        memory.save_context(
            inputs={"input": modified_messages[0].content},
            outputs={"output": response.content}
        )
        return {'messages': [response]}

    return {'messages': [response]}

# Router node logic to determine the next step
def router_node(state: MessagesState) -> str:
    last_message = state['messages'][-1]
    if last_message.tool_calls:
        return 'continue'
    return 'respond'

# Define the workflow
workflow = StateGraph(MessagesState)
workflow.add_node("llm", call_model)
workflow.add_node("tools", ToolNode(tools))
workflow.set_entry_point("llm")
workflow.add_conditional_edges(
    "llm",
    router_node,
    {
        "respond": END,
        "continue": "tools"
    }
)
workflow.add_edge("tools", "llm")

# Compile the app
app = workflow.compile()
```

```python
import os
import httpx
import requests
from debugging.logger import logging
from dotenv import load_dotenv

load_dotenv()

WHATSAPP_API_URL = os.getenv('WHATSAPP_API_URL')
WHATSAPP_ACCESS_TOKEN = os.getenv("WHATSAPP_ACCESS_TOKEN")
PHONE_NUMBER_ID = os.getenv("PHONE_NUMBER_ID")

headers = {
    "Authorization": f"Bearer {WHATSAPP_ACCESS_TOKEN}",
    "Content-Type": "application/json"
}

async def send_text_message(to: str, message: str)->None:
    """
    Send a plain text message to a user via WhatsApp Cloud API.
    """
    url = f"{WHATSAPP_API_URL}/{PHONE_NUMBER_ID}/messages"
    payload = {
        "messaging_product": "whatsapp",
        "to": to,
        "type": "text",
        "text": {"body": message}
    }
    logging.info(f"Sending text message to {to}: {message}")
    async with httpx.AsyncClient() as client:
        response = await client.post(url, headers=headers, json=payload)
        logging.info(f"Text message sent to {to}. Status: {response.status_code}")

async def upload_media(image_url: str) -> str:
    # 1) Fetch the image bytes synchronously
    resp = requests.get(image_url)
    resp.raise_for_status()
    image_bytes  = resp.content
    content_type = resp.headers.get("Content-Type", "application/octet-stream")

    # 2) Prepare multipart file tuple: (filename, bytes, content_type)
    filename = os.path.basename(image_url).split("?")[0] or "upload"
    files = {
        "file": (filename, image_bytes, content_type)
    }

    # 3) Upload to WhatsApp
    url    = f"{WHATSAPP_API_URL}/{PHONE_NUMBER_ID}/media"
    params = {"messaging_product": "whatsapp"}
    headers= {"Authorization": f"Bearer {WHATSAPP_ACCESS_TOKEN}"}

    async with httpx.AsyncClient() as client:
        r = await client.post(url, headers=headers, params=params, files=files)
        r.raise_for_status()
        media_id = r.json().get("id")
        logging.info(f"Uploaded media, got media_id={media_id}")
        return media_id        


async def send_image_by_id(to: str, media_id: str, caption: str="") -> None:
    payload = {
        "messaging_product": "whatsapp",
        "to": to,
        "type": "image",
        "image": {"id": media_id, "caption": caption}
    }
    async with httpx.AsyncClient() as client:
        r = await client.post(f"{WHATSAPP_API_URL}/{PHONE_NUMBER_ID}/messages", headers=headers, json=payload)
        logging.info(f"Image-by-ID sent, status {r.status_code}")
```

```python
import os
import requests
from base64 import b64encode
from debugging.logger import logging

# Set your WhatsApp API access token and ImgBB API key
ACCESS_TOKEN= os.getenv('WHATSAPP_ACCESS_TOKEN')
IMGBB_API_KEY = os.getenv('IMGBB_API_KEY')
IMGBB_URL=os.getenv("IMGBB_URL")

def _get_whatsapp_media_url(media_id: str) -> str:
    """
    Get the temporary media URL from WhatsApp using the media ID.
    """
    url = f"https://graph.facebook.com/v22.0/{media_id}"
    headers = {"Authorization": f"Bearer {ACCESS_TOKEN}"}
    res = requests.get(url, headers=headers)
    return res.json().get("url")

def _get_image_bytes(media_url: str) -> bytes:
    """
    Download the image bytes from the media URL using WhatsApp authorization.
    """
    headers = {"Authorization": f"Bearer {ACCESS_TOKEN}"}
    res = requests.get(media_url, headers=headers)
    return res.content

def whatsapp_image_to_https(media_id: str) -> str:
    """
    Converts a WhatsApp media ID to a publicly accessible HTTPS image URL
    by uploading it to ImgBB.
    """
    # Step 1: Get the temporary media URL from WhatsApp
    media_url = _get_whatsapp_media_url(media_id)

    # Step 2: Download the image bytes from WhatsApp
    image_bytes = _get_image_bytes(media_url)

    # Step 3: Upload image to ImgBB and get the HTTPS URL
    payload = {
        'key': IMGBB_API_KEY,
        'image': b64encode(image_bytes),
        'name': 'uploaded_image',
        "expiration": 0
    }

    response = requests.post(IMGBB_URL, data=payload)

    # Step 4: Return the image URL if upload is successful
    if response.status_code == 200:
        logging.info("New image url successfully converted to https link")
        return response.json()['data']['display_url']
    else:
        logging.info("New image url threw error while converting into https link")
        raise Exception(f"Failed to upload to ImgBB: {response.content}")
```

```python
import os
import httpx
import asyncio
import sys
from debugging.logger import logging
from debugging.exception import customException
from dotenv import load_dotenv
import requests

load_dotenv()

WHATSAPP_API_URL       = os.getenv("WHATSAPP_API_URL")
WHATSAPP_ACCESS_TOKEN  = os.getenv("WHATSAPP_ACCESS_TOKEN")
PHONE_NUMBER_ID        = os.getenv("PHONE_NUMBER_ID")

# Shared headers & timeouts
HEADERS = {
    "Authorization": f"Bearer {WHATSAPP_ACCESS_TOKEN}",
    "Content-Type": "application/json"
}
TIMEOUT = httpx.Timeout(
    connect=10.0,
    read=30.0,
    write=10.0,
    pool=5.0
)
MAX_RETRIES = 5
RETRY_DELAY = 2  # seconds
```

```python
def send_image_by_id(to: str, media_id: str, caption: str = "") -> bool:
    """
    Send an already‐uploaded image via its `media_id`.
    Returns True on success, False otherwise.
    """
    url = f"{WHATSAPP_API_URL}/{PHONE_NUMBER_ID}/messages"
    payload = {
        "messaging_product": "whatsapp",
        "to": to,
        "type": "image",
        "image": {"id": media_id, "caption": caption}
    }
    for attempt in range(1, MAX_RETRIES + 1):
        try:
            logging.info(f"[Attempt {attempt}] Sending image-id {media_id} to {to}")
            r = requests.post(url, headers=HEADERS, json=payload)
            r.raise_for_status()
            logging.info(f"Image-by-ID sent (status {r.status_code})")
            return True,r
        except (httpx.ConnectTimeout, httpx.ReadTimeout) as e:
            logging.warning(f"Timeout on attempt {attempt} sending image-id: {e!r}")
        except httpx.HTTPStatusError as e:
            status = e.response.status_code
            if 500 <= status < 600 and attempt < MAX_RETRIES:
                logging.warning(f"Server error {status} sending image-id; retrying...")
            else:
                logging.error(f"Failed to send image-by-id (status {status}): {e.response.text}")
                break
        except Exception as e:
            logging.error(f"Unexpected error sending image-by-id: {e!r}")
            break

    logging.error(f"All retries failed for sending image-id {media_id} to {to}")
    return False
```

```python
import sys
import warnings

# Ignore all warnings
warnings.filterwarnings("ignore")

from dotenv import load_dotenv
from fastapi import FastAPI, Request, Query, Response, HTTPException

from Whatsapp.receive_whatsapp import handle_whatsapp_message
from Whatsapp.send_whatsapp import send_text_message
from Whatsapp.utils.spamming import is_spamming, is_repeating_same_message, is_blocked

from debugging.logger import logging
from debugging.exception import customException

# Load environment variables
load_dotenv()

from cachetools import TTLCache
from collections import defaultdict
import asyncio

# initialization
app = FastAPI()
PROCESSED_MESSAGES = TTLCache(maxsize=1000, ttl=300)  # 5-minute cache
USER_PROCESSING_STATE = defaultdict(lambda: False)
USER_QUEUES = defaultdict(asyncio.Queue)

VERIFY_TOKEN = "my_custom_verify_token"
if not VERIFY_TOKEN:
    logging.error("VERIFY_TOKEN is not set in environment variables.")
    raise RuntimeError("VERIFY_TOKEN must be set in environment variables.")

@app.api_route("/webhook", methods=["GET", "POST"])
async def webhook(
    request: Request,
    hub_mode: str = Query(default=None, alias="hub.mode"),
    hub_challenge: str = Query(default=None, alias="hub.challenge"),
    hub_verify_token: str = Query(default=None, alias="hub.verify_token")
):
    try:
        # ─── Verification (GET) ───
        if request.method == "GET":
            logging.info("GET request received for webhook verification.")
            if hub_mode == "subscribe" and hub_verify_token == VERIFY_TOKEN:
                logging.info("Webhook verified successfully.")
                return Response(content=hub_challenge, media_type="text/plain")
            logging.warning("Webhook verification failed.")
            return Response(status_code=403, content="Forbidden")

        # ─── Incoming payload (POST) ───
        elif request.method == "POST":
            payload = await request.json()
            logging.info("Received POST data")
            # ─── 1) Short-circuit status updates ───
            entry = payload.get("entry", [{}])[0]
            change = entry.get("changes", [{}])[0]
            value = change.get("value", {})

            statuses = value.get("statuses")
            if statuses:
                for status in statuses:
                    logging.info(
                        f"Status update for {status.get('recipient_id')}: "
                        f"{status.get('status')} at {status.get('timestamp')}"
                    )
                # Don't process as a message
                return {"status": "status_update"}

            # ─── 2) Extract message ID and check duplicates ───
            try:
                message = value.get("messages", [])[0]
                message_id = message["id"]
                phone_number = message["from"]
                
                # Check if we've already processed this message
                if message_id in PROCESSED_MESSAGES:
                    logging.info(f"Ignoring duplicate message: {message_id}")
                    return {"status": "duplicate"}
                
                # Add to processed cache
                PROCESSED_MESSAGES[message_id] = True
                await USER_QUEUES[phone_number].put((message,payload)) # Store message

                # Start processing if not already
                if not USER_PROCESSING_STATE[phone_number]:
                    asyncio.create_task(process_user_queue(phone_number))

                return {"status": "queued"}
            
            except Exception as e:
                logging.error(f"Invalid message: {e}")
                return {"status": "ignored"}

        else:
            raise HTTPException(status_code=405, detail="Method Not Allowed")

    except Exception as e:
        logging.error(str(customException(e, sys)))
        if phone_number:
            try:
                await send_text_message(to=phone_number, message="Server busy. Try again later.")
            except Exception as inner:
                logging.error(f"Notify failed: {inner}")
        raise HTTPException(status_code=500, detail="Internal Server Error")   

# ────────────────────────────────────────────────────────────────
# QUEUE HANDLER FUNCTION
# ────────────────────────────────────────────────────────────────                     
async def process_user_queue(phone_number):
    USER_PROCESSING_STATE[phone_number] = True
    try:
        while not USER_QUEUES[phone_number].empty():
            message, payload = await USER_QUEUES[phone_number].get()

            message_type = message.get("type")
            text = message.get("text", {}).get("body", "").strip()     

            # Spam & duplicate checks
            if message_type == "text" and text:
                if await is_blocked(phone_number):
                    logging.info(f"Blocked user tried to message: {phone_number}")
                    await send_text_message(to=phone_number, message="You are blocked for 20 minutes. Try again later.")
                    continue

                if await is_spamming(phone_number):
                    await send_text_message(to=phone_number, message="You are blocked for 20 minutes due to spamming.")
                    continue

                if await is_repeating_same_message(phone_number, text):
                    await send_text_message(to=phone_number, message="Don't send the same message again and again.")
                    
            try:       
                # ─── 4) Hand off to your existing handler ───
                result = await handle_whatsapp_message(data=payload)
                if isinstance(result, dict) and result.get("error"):
                    await send_text_message(to=phone_number, message="server busy. Try again.")
            except Exception as handler_err:
                logging.error(f"Handler error for {phone_number}: {handler_err}")
                await send_text_message(to=phone_number, message="Something went wrong. Try again.")
            finally:
                await asyncio.sleep(0.1)  # small delay before processing next        

    finally:
        USER_PROCESSING_STATE[phone_number] = False
```
